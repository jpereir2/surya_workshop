### Speed check: is the bottleneck S3 or the GPU?

With `s3_mode: download`, each sample pulls **three** NetCDF files (~0.5 GB each) from S3 the
first time it is visited, then reads them from `s3_cache_dir` forever after. When that happens
*inside* the training loop, the GPU idles through the download — which is what a flat WandB
GPU-utilization trace with one brief spike per step looks like.

The cell below walks each dataset twice. **Pass 1 is cold** (downloads, and leaves the files
cached). **Pass 2 is warm** (local disk only). Two things come out of it:

* The cache is now warm, so the training run below is not competing with the network.
* The cold/warm ratio tells you which resource you are actually short of, which decides how
  much data the real runs can afford.

````python
# Walk every sample twice: pass 1 downloads and caches, pass 2 reads from local disk.
# Each sample is loaded, normalized, then discarded, so only one (~2.6 GB) is held at a time.
import time

def walk(ds, label):
    per = []
    for i in range(len(ds)):
        t = time.perf_counter()
        _ = ds[i]                              # load + normalize, then let it go
        per.append(time.perf_counter() - t)
        print(f"\r  {label}: {i+1}/{len(ds)}   last {per[-1]:6.1f}s   "
              f"mean {sum(per)/len(per):6.1f}s", end="", flush=True)
    print()
    return sum(per) / len(per)

for _name, _dl in (("train", train_data_loader), ("val", val_data_loader)):
    _cold = walk(_dl.dataset, f"{_name} cold")
    _warm = walk(_dl.dataset, f"{_name} warm")
    print(f"  -> {_name}: {_cold:.1f}s/sample cold, {_warm:.1f}s/sample warm "
          f"({_cold/max(_warm, 1e-9):.1f}x faster once cached)\n")

print("If warm is much faster than cold, S3 was the bottleneck: warm the cache for every")
print("sample you plan to use BEFORE training, and num_workers=2 then becomes safe.")
print("If warm is about the same, the GPU is the limit: shorten model.checkpoint_layers.")
````

### Load the image decoder that `load_pretrained_weights()` misses

The cell above reports **157 / 159**. The two missing tensors are the model's *final layer* —
the one that converts Surya's internal representation (65,536 patches x 1,280 numbers) back
into a 13-channel image.

The checkpoint stores that layer as `unembed.*`. Our model calls it `head_unembed.*`, because
`apply_peft_lora()` finds trainable head modules by the `head_` prefix. But
`load_pretrained_weights()` only tries the key as-is and with a `backbone.` prefix
(`workshop_infrastructure/utils.py:404`), so these two are dropped **without a warning** and
the layer starts as random numbers.

That is why a prediction looks foggy and dim, and why the loss parks near 1.0: in normalized
space every channel has mean 0 and standard deviation 1, so "predict the average and give up"
scores almost exactly 1.0. The body of the model knows the Sun; the last step was guessing.

**This cell must run before the LoRA cell** — `apply_peft_lora()` rewrites parameter names.

````python
# Copy the two decoder tensors load_pretrained_weights() could not find, by trying the
# "head_" prefix it never tries. Asserts rather than warns: a silent miss here is exactly
# the failure mode we are fixing.
_ckpt = torch.load(cfg.model.pretrained_path, weights_only=True, map_location="cpu")
_msd = model.state_dict()

_copied = []
for _k, _v in _ckpt.items():
    _target = f"head_{_k}"          # "unembed.unembed.0.weight" -> "head_unembed.unembed.0.weight"
    if _target in _msd:
        assert _v.shape == _msd[_target].shape, (
            f"shape mismatch for {_target}: checkpoint {tuple(_v.shape)} vs "
            f"model {tuple(_msd[_target].shape)}. Check ft_out_chans == len(cfg.data.channels)."
        )
        _msd[_target] = _v
        _copied.append((_k, _target, tuple(_v.shape)))

assert _copied, (
    "No checkpoint key matched head_*. Expected 'unembed.unembed.0.weight'. "
    f"Checkpoint keys beginning with 'unembed': {[k for k in _ckpt if k.startswith('unembed')]}"
)

model.load_state_dict(_msd, strict=True)
print(f"Copied {len(_copied)} decoder tensors the loader missed:")
for _k, _target, _shape in _copied:
    print(f"    {_k}  ->  {_target}   {_shape}")
print("\n3328 = 16*16 patch pixels x 13 channels, so this really is our LinearDecoder.")
del _ckpt, _msd
````

## To LoRA or not to Lora

This cell gives you two options.  On the one hand we have the classic freezing of the backbone (the initial layers of the model).   On the other hand we have the use of a LoRA.

LoRas have been a remarkable addition to our arsenal of models.   They have the advantage of keeping pretty much the entire model intact and only add broad modifications to weights as needed.

````python
# Three fine-tuning regimes, all selected from the model: section of the config:
#
#   use_lora: true                          -> LoRA adapters on attention + FFN (default)
#   use_lora: false, freeze_backbone: true  -> linear probe: only the head trains
#   use_lora: false, freeze_backbone: false -> full fine-tuning of all 366M parameters
#
# 3_finetune_template_1D.py applies exactly this logic in build_model().
if cfg.model.freeze_backbone:
    for name, param in model.named_parameters():
        if name.startswith("backbone."):
            param.requires_grad = False

if cfg.model.use_lora:
    model = apply_peft_lora(model, cfg.model.lora_config)

trainable = sum(p.numel() for p in model.parameters() if p.requires_grad)
total = sum(p.numel() for p in model.parameters())
print(f"Trainable parameters: {trainable:,} / {total:,} ({100 * trainable / total:.2f}%)")
````

### Forced flip pass: does a flipped stack reach the model, and does it cost more?

This is the plumbing check for the whole project. It takes **one** sample and pushes it
through the model twice — upright, and vertically flipped the same way `flare_dataset.py`
flips it (`axis=-2`, i.e. north-south). It answers:

1. **Does the flip actually reach the model?** Asserted two ways: flipping twice must return
   the original, and the flipped input must differ from the upright one.
2. **Does a flipped input take longer?** It should not. The model performs identical
   arithmetic no matter what the pixel values are — no data-dependent branching, no shortcut
   for dark pixels. A ratio near **1.00** is the correct result, not a broken test.
3. **How much does the flip change the model's answer?** Un-flip the flipped prediction and
   compare it to the upright prediction. Near 0 means the model is blind to orientation, so
   Hale's law cannot explain a cycle-25 failure. Large means its answer genuinely depends on
   orientation — your hypothesis is live. This is the single number the project turns on, and
   you get it here for free, before any training.
4. **What does one real training step cost?** Forward + backward + optimizer at the current
   `checkpoint_layers`. Every time estimate for the overnight runs rests on this number, so
   measure it rather than trusting an estimate.

It puts the model back on the CPU at the end, so nothing downstream is affected. Cell 23's
CPU forward pass becomes redundant once this runs — skip it.

````python
# Force one upright and one flipped pass through the model, then time a real training step.
import time

dev = "cuda" if torch.cuda.is_available() else "cpu"
chk_batch = next(iter(train_data_loader))

# The backbone reads only these three keys (helio_spectformer.py:258-269), so moving just
# them keeps the 0.9 GB forecast tensor off the GPU during the forward-only tests.
_KEYS = ("ts", "time_delta_input", "lead_time_delta")
up = {k: chk_batch[k][:1].to(dev) for k in _KEYS}
target_up = chk_batch["forecast"][:1, :, 0].float().to(dev)   # (1, 13, 4096, 4096)

# The same vertical flip flare_dataset.py:151 applies. ts is (B, C, T, H, W), so -2 is H.
_t = time.perf_counter()
fl = dict(up)
fl["ts"] = torch.flip(up["ts"], dims=[-2])
_flip_ms = (time.perf_counter() - _t) * 1e3

assert torch.equal(torch.flip(fl["ts"], dims=[-2]), up["ts"]), "flip is not its own inverse"
assert not torch.equal(fl["ts"], up["ts"]), "flipped input equals upright — the flip did nothing"
print(f"flip verified (is its own inverse, and does change the data); cost {_flip_ms:.0f} ms\n")

model = model.to(dev).eval()

def timed(b, label, reps=2):
    with torch.no_grad(), torch.autocast(device_type=dev, dtype=torch.bfloat16):
        out = model(b)                                   # warm-up: first call has one-off cost
        if dev == "cuda": torch.cuda.synchronize()
        best = float("inf")
        for _ in range(reps):
            _t = time.perf_counter()
            out = model(b)
            if dev == "cuda": torch.cuda.synchronize()
            best = min(best, time.perf_counter() - _t)
    print(f"  {label:8s} forward {best:6.2f} s    output {tuple(out.shape)}")
    return best, out.float()

t_up, pred_up = timed(up, "upright")
t_fl, pred_fl = timed(fl, "flipped")

_mse = lambda a, b: torch.mean((a - b) ** 2).item()
print(f"\nflipped / upright time ratio           : {t_fl / t_up:.3f}    (expect ~1.0)")
print(f"error vs ground truth, upright input   : {_mse(pred_up, target_up):.4f}")
print(f"error of just copying the last input   : {_mse(up['ts'][:, :, -1].float(), target_up):.4f}"
      "   <- persistence baseline: beat this or there is no forecast skill")
print(f"how much the flip changed the answer   : {_mse(torch.flip(pred_fl, dims=[-2]), pred_up):.4f}"
      "   <- ~0: model is blind to orientation; large: orientation matters")

# One real training step at the current checkpoint_layers setting.
model.train()
_opt = torch.optim.Adam([p for p in model.parameters() if p.requires_grad], lr=1e-4)
if dev == "cuda": torch.cuda.reset_peak_memory_stats()
_steps = []
for _ in range(2):
    _t = time.perf_counter()
    with torch.autocast(device_type=dev, dtype=torch.bfloat16):
        _loss = torch.nn.functional.mse_loss(model(up).float(), target_up)
    _loss.backward()
    _opt.step()
    _opt.zero_grad(set_to_none=True)
    if dev == "cuda": torch.cuda.synchronize()
    _steps.append(time.perf_counter() - _t)

_step = min(_steps)
print(f"\ntraining step, batch 1 : {_step:.1f} s  ->  {3600/_step:.0f} sample-views per GPU-hour")
print(f"                        so a 150-sample x 2-epoch run is about {300*_step/3600:.1f} hours")
if dev == "cuda":
    _peak = torch.cuda.max_memory_allocated() / 2**30
    _tot = torch.cuda.get_device_properties(0).total_memory / 2**30
    print(f"peak GPU memory        : {_peak:.1f} GB of {_tot:.0f} GB")
    print(f"model.checkpoint_layers: {cfg.model.checkpoint_layers}")
    print("  Those layers recompute their activations during backward to save memory, at"
          " roughly 30-40% extra compute.")
    print(f"  You are using {_peak:.0f} of {_tot:.0f} GB, so shortening that list (try [0,1,2,3,4])"
          " should give faster steps for free.")

# Leave the notebook exactly as it was, so cell order does not matter.
del _opt, _loss, pred_up, pred_fl, up, fl, target_up, chk_batch
model = model.to("cpu").train()
if dev == "cuda": torch.cuda.empty_cache()
print("\nModel returned to CPU; Lightning will move it back to the GPU during fit().")
````

### The small check: flips through the real pipeline, before any long run

Three questions, in order, on a handful of samples.

**A. Does the dataset's flip fire, and does it flip the *target* too?** This is the one that
matters most. If the input were flipped but the ground-truth image were not, you would be asking
the model to predict a mirror image of what it was shown — guaranteed nonsense, and the loss would
look bad for a reason that has nothing to do with Hale's law. Asserted directly.

**B. Does orientation change the answer?** Each sample is scored **twice** — once upright, once
flipped — with both its input and its target flipped together. Same flare, both orientations, so
any difference is orientation and not "which samples happened to get flipped." That pairing is what
makes a 6-sample check readable; with random flipping you would need hundreds of samples to see the
same thing. Random mixing is still the right thing for the real training runs — this is only how
you measure precisely when N is tiny.

Each row also carries the **persistence** error: what you'd score by just copying the last input
frame forward. It doesn't involve the model, and flipping cannot change it, so upright and flipped
persistence must come out **identical**. If they don't, the flip is being applied inconsistently
somewhere and nothing else in the table can be trusted.

**C. Do the flips actually reach the training loop?** Turns the random flipping back on, pulls real
batches through `train_data_loader`, prints `was_flipped` per batch, and takes a real optimizer
step. This is the end-to-end confirmation.

Cost is roughly two forward passes per sample plus one read each, so it scales with
`max_number_of_samples` in cell 11. Six samples is plenty.

````python
# Small end-to-end flip check. Runs at batch size 1 throughout to stay well clear of memory limits.
# Set RUN_TRAINING_STEP = False for the minimum check: parts A and B only, no optimizer.
RUN_TRAINING_STEP = True
import numpy as np
import pandas as pd
import torch

dev = "cuda" if torch.cuda.is_available() else "cpu"
_KEYS = ("ts", "time_delta_input", "lead_time_delta")   # the only keys the backbone reads
mse = lambda a, b: torch.mean((a.float() - b.float()) ** 2).item()

def as_batch(sample):
    """One dataset sample -> a batch of 1 on the device, target separated out."""
    b = {k: torch.as_tensor(sample[k])[None].to(dev) for k in _KEYS}
    tgt = torch.as_tensor(sample["forecast"])[None][:, :, 0].to(dev)   # (1, 13, 4096, 4096)
    return b, tgt

# ---------------------------------------------------------------- A. the dataset's own flip
ds = train_data_loader.dataset
saved_p = ds.flip_probability

ds.flip_probability = 0.0
s_up = ds[0]
ds.flip_probability = 1.0
s_fl = ds[0]

assert not s_up["was_flipped"], "p=0.0 still reported was_flipped=True"
assert s_fl["was_flipped"], "p=1.0 did not report was_flipped=True"
assert np.array_equal(s_fl["ts"], np.flip(s_up["ts"], axis=-2)), \
    "input was not flipped north-south the way we expect"
assert np.array_equal(s_fl["forecast"], np.flip(s_up["forecast"], axis=-2)), \
    "TARGET was not flipped with the input -- the model would be asked to predict a mirror image"
print("A. flip fires, is recorded in was_flipped, and flips BOTH input and target. OK\n")
del s_up, s_fl

# ---------------------------------------------------------------- B. paired scoring, both splits
model = model.to(dev).eval()
rows = []
for split, dl in (("train", train_data_loader), ("val", val_data_loader)):
    d = dl.dataset
    d.flip_probability = 0.0            # we apply the flip ourselves, so the pairing is exact
    for i in range(len(d)):
        s = d[i]
        b, tgt = as_batch(s)
        for orient in ("upright", "flipped"):
            bb, t = dict(b), tgt
            if orient == "flipped":
                bb["ts"] = torch.flip(b["ts"], dims=[-2])
                t = torch.flip(tgt, dims=[-2])
            with torch.no_grad(), torch.autocast(device_type=dev, dtype=torch.bfloat16):
                pred = model(bb).float()
            rows.append({"split": split, "i": i, "time": s["ds_index"][:16],
                         "orientation": orient, "error": mse(pred, t),
                         "persistence": mse(bb["ts"][:, :, -1], t)})
            print(f"\r  B: {split} {i+1}/{len(d)} {orient:8s}", end="", flush=True)
        del b, tgt, bb, t, pred
print()

# Part B borrowed both datasets by zeroing flip_probability; give it back immediately, so
# anything run afterwards -- the plotting cells especially -- sees the flips again.
for d in (train_data_loader.dataset, val_data_loader.dataset):
    d.flip_probability = saved_p
print(f"flip_probability restored to {saved_p} on both datasets")

tab = pd.DataFrame(rows)
print(tab.to_string(index=False), "\n")
print(tab.groupby(["split", "orientation"])[["error", "persistence"]].mean().to_string(), "\n")

# Persistence never touches the model and a flip cannot change it, so these must agree exactly.
_p = tab.pivot_table(index=["split", "i"], columns="orientation", values="persistence")
_gap = (_p["upright"] - _p["flipped"]).abs().max()
assert _gap < 1e-4, f"persistence differs by {_gap:.2e} between orientations -- flip is inconsistent"
print(f"persistence matches across orientations (max gap {_gap:.2e}). Table is trustworthy.")

_m = tab.pivot_table(index=["split", "i"], columns="orientation", values="error")
_delta = (_m["flipped"] - _m["upright"]).mean()
print(f"mean error, flipped minus upright: {_delta:+.4f}")
print("  ~0 -> orientation does not matter, so Hale's law cannot explain the cycle-25 failure.")
print("  clearly positive -> the model is tied to the orientation it was pretrained on.")

if not RUN_TRAINING_STEP:
    model = model.to("cpu").train()
    if dev == "cuda":
        torch.cuda.empty_cache()
    raise SystemExit("Parts A and B done; part C skipped. `tab` holds the table, "
                     f"flip_probability is back at {saved_p}.")

# ---------------------------------------------------------------- C. flips reach the training loop
model.train()
opt = torch.optim.Adam([p for p in model.parameters() if p.requires_grad], lr=1e-4)
for d in (train_data_loader.dataset, val_data_loader.dataset):
    d.flip_probability = saved_p

print(f"\nC. two batches through train_data_loader at flip_probability="
      f"{train_data_loader.dataset.flip_probability}:")
for n, batch in enumerate(train_data_loader):
    if n == 2:
        break
    flags = [bool(f) for f in batch["was_flipped"]]
    b = {k: batch[k][:1].to(dev) for k in _KEYS}            # model sees 1 sample; flags cover all
    t = batch["forecast"][:1][:, :, 0].to(dev)
    with torch.autocast(device_type=dev, dtype=torch.bfloat16):
        loss = torch.nn.functional.mse_loss(model(b).float(), t.float())
    loss.backward()
    opt.step()
    opt.zero_grad(set_to_none=True)
    print(f"   batch {n}: was_flipped={flags}  loss={loss.item():.4f}  (stepped)")
    del b, t, loss

print("\nFlips survive the dataset, the collate, and a real optimizer step.")
if dev == "cuda":
    print(f"peak GPU memory this cell: {torch.cuda.max_memory_allocated()/2**30:.1f} GB")

del opt, rows
model = model.to("cpu").train()
if dev == "cuda":
    torch.cuda.empty_cache()
print("Model back on CPU. `tab` is kept so you can re-inspect it.")
````

## Define your PyTorch ligthning module

In this workshop we will use PyTorch lightning to train our models.  PyTorch lighting reduces the amount of code required to implement a training loop in comparison to PyTorch (at the expense of control and versatility).  

Opening the FlareLightningModule shows a simple Lightning model implementation.  It consists of:

- An initialization of the class (metrics, model, and learning rate).
- The forward code that runs evaluation of the model.
- Training and validation steps.
- Configuration of optimizers.

**_Note that it is the same Lightning module we used for the baseline!!_**

````python
from downstream_apps.flare_forecast.lightning_modules.pl_flare_baseline import FlareLightningModule
````

````python
class Flare2DLightningModule(FlareLightningModule):
    """FlareLightningModule with a 2D image target instead of a scalar label."""

    @staticmethod
    def _target(batch):
        fc = batch["forecast"]
        assert fc.shape[2] == 1, (
            f"expected one target frame, got {fc.shape[2]}. Set training.rollout_steps: 0."
        )
        return fc[:, :, 0].float()          # (B, C, H, W)

    def _step(self, batch, stage):
        target = self._target(batch)
        if self.preprocess_fn is not None:
            batch = self.preprocess_fn(batch)
        output = self(batch)

        loss_fn = self.training_loss if stage == "train" else self.validation_loss
        eval_fn = self.training_evaluation if stage == "train" else self.validation_evaluation

        losses, weights = loss_fn(output, target)
        loss = self._combine_losses(losses, weights)
        self.log(f"{stage}_loss", loss, prog_bar=True, batch_size=self.batch_size, sync_dist=True)
        for k, v in losses.items():
            self.log(f"{stage}_loss_{k}", v, batch_size=self.batch_size, sync_dist=True)

        mets, mw = eval_fn(output, target)
        if len(mw) > 0:
            for k, v in mets.items():
                self.log(f"{stage}_metric_{k}", v, batch_size=self.batch_size, sync_dist=True)
        return loss

    def training_step(self, batch, batch_idx):   return self._step(batch, "train")
    def validation_step(self, batch, batch_idx): self._step(batch, "val")
````

````python
import numpy as np
import pandas as pd
import torch
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
import sunpy.visualization.colormaps as cm  # registers 'sdoaia131', 'hmimag', ...
from torch.utils.data._utils.collate import default_collate

# ==================================================================
# CONFIG FOR THIS CHECK
# ==================================================================
CHANNELS_TO_PLOT = ["aia131", "aia171"]          # one row per channel
SCALE_VERSIONS   = [("normal", 1), ("enhanced", 10)]   # 99th-pct multiplier
STRIDE           = 1     # 1 = full 4096. Set 4 to make the figure ~16x faster;
                         # inverse_transform is elementwise, so striding is exact.
SHARE_GT_PRED_SCALE = True   # see note under section 4

train_dataset = train_data_loader.dataset        # nb0's name for it
idx           = 0

channel_order = cfg.data.channels
ch_idx = {c: channel_order.index(c) for c in CHANNELS_TO_PLOT}

item = train_dataset[idx]

n_inputs = item["ts"].shape[1]
n_lead   = item["forecast"].shape[1]
assert n_lead == 1, f"this plot assumes one target frame; got {n_lead}. Set rollout_steps: 0."

# ==================================================================
# 0. RUN THE MODEL  (NEW)
# ==================================================================
# lit_model, not model: lit_model wraps the PEFT model with the trained adapters
# and head. Lightning moves it back to CPU after fit, hence the explicit .to().
lit_model = lit_model.to("cuda").eval()
with torch.no_grad(), torch.autocast("cuda", dtype=torch.bfloat16):
    _b = default_collate([item])
    _b = {k: (v.to("cuda") if torch.is_tensor(v) else v) for k, v in _b.items()}
    pred_norm = lit_model(_b)[0].float().cpu().numpy()   # (13, H, W), normalized space
del _b; torch.cuda.empty_cache()

# ==================================================================
# 1. TRUE FRAME TIMESTAMPS  (from the sample itself, not from ds_index)
# ==================================================================
reference_timestep = pd.Timestamp(train_dataset.valid_indices[idx])

# helio.py: the LAST entry of the sorted input offsets is the "now" frame.
now_offset_min = sorted(cfg.data.time_delta_input_minutes)[-1]
now_time = reference_timestep + pd.Timedelta(minutes=now_offset_min)

# time_delta_input = hours BEFORE now ; lead_time_delta = hours before now (neg = future)
input_times  = [now_time - pd.Timedelta(minutes=round(float(h) * 60))
                for h in item["time_delta_input"]]
target_times = [now_time - pd.Timedelta(minutes=round(float(h) * 60))
                for h in item["lead_time_delta"]]

# ==================================================================
# 2. PROVENANCE REPORT
# ==================================================================
row = train_dataset.df_valid_indices.iloc[idx]
anchor_col   = cfg.data.ds_time_column
anchor_event = pd.Timestamp(item["ds_index"])
gt_time      = target_times[0]
gap          = abs(gt_time - anchor_event)

print(f"FLARE EVENT      : {row['GOES_class']}   ")
print(f"  start_time     : {row['start_time']}")
print(f"  peak_time      : {row['peak_time']}")
print(f"  end_time       : {row['end_time']}")
print(' ')
print(' ')
print(f"ds_time_column = '{anchor_col}'  ->  {anchor_event}")
print(' ')
print(f"Ground Truth {gt_time}")
print(f"Reference Timestamp (helio.py anchor): {reference_timestep}")
print(f"    offset from {anchor_col:<11}: {gap}  "
      f"({gap.total_seconds()/60:.0f} min)")
print(f"    ds_index vs reference gap: {abs(reference_timestep - anchor_event)}")
print(' ')
print(' ')
print("config: ", cfg.data.time_delta_input_minutes,
      f"  | n_input_timestamps = {train_dataset.n_input_timestamps}")
print(' ')
print("source files:")
for k, ts in enumerate(input_times):
    print(f"  [input {k}] {ts}  (t{'' if k==0 else ''}={(ts-now_time).total_seconds()/60:+.0f} min from now)"
          f"\n            -> {train_dataset.index.loc[ts, 'path']}")
for k, ts in enumerate(target_times):
    print(f"  [GT    {k}] {ts}  ({(ts-now_time).total_seconds()/60:+.0f} min from now)"
          f"\n            -> {train_dataset.index.loc[ts, 'path']}")
print(' ')
print(f"Flipped?  ->  flare_dataset = {item.get('was_flipped')} "
      f"| helio = {getattr(train_dataset, 'random_vert_flip', None)}")
print(' ')

# ==================================================================
# 2b. PREDICTION QUALITY  (NEW)
# ==================================================================
# Per-channel means, then averaged: identical to the global mean (equal sizes) but
# holds only one 67 MB channel at a time instead of four 872 MB arrays.
_gt_n  = item["forecast"][:, 0]      # normalized space -- what the loss actually sees
_now_n = item["ts"][:, -1]           # persistence: copy the "now" frame forward
mse_model = float(np.mean([((pred_norm[k] - _gt_n[k]) ** 2).mean()
                           for k in range(len(channel_order))]))
mse_persist = float(np.mean([((_now_n[k] - _gt_n[k]) ** 2).mean()
                             for k in range(len(channel_order))]))
print("NORMALIZED-SPACE MSE  (all 13 channels -- comparable to train_loss/val_loss)")
print(f"  model              : {mse_model:.5f}")
print(f"  persistence        : {mse_persist:.5f}   <- copy 'now' forward, no model")
print(f"  skill vs persist.  : {1 - mse_model/mse_persist:+.1%}   (>0 = model helps)")
print(' ')
print("PER-CHANNEL RMSE, physical units")
print(f"  {'channel':<10} {'model':>12} {'persistence':>14}")
for c, n in ch_idx.items():
    rm = float(np.sqrt(((pred_norm[n] - _gt_n[n]) ** 2).mean()))
    rp = float(np.sqrt(((_now_n[n]    - _gt_n[n]) ** 2).mean()))
    print(f"  {c:<10} {rm:>12.4g} {rp:>14.4g}")
print(' ')
print(' ')

# ==================================================================
# 3. PHYSICAL-UNIT FRAMES PER CHANNEL
# ==================================================================
# Transform each FRAME once (all 13 channels at a time -- inverse_transform_data
# indexes per-channel stats, so the channel axis must stay whole), then slice the
# channels we plot out of the result.
inv, _c, S = train_dataset.inverse_transform_data, np.ascontiguousarray, STRIDE
ins_phys  = [inv(_c(item["ts"][:, t, ::S, ::S]))       for t in range(n_inputs)]
tgts_phys = [inv(_c(item["forecast"][:, l, ::S, ::S])) for l in range(n_lead)]
prds_phys = [inv(_c(pred_norm[:, ::S, ::S]))]

frames = {c: ([a[n] for a in ins_phys],
              [a[n] for a in tgts_phys],
              [a[n] for a in prds_phys]) for c, n in ch_idx.items()}

# ==================================================================
# 4. PLOTTING
# ==================================================================
def _lim(a, factor):
    nz = a[a != 0]
    return (np.percentile(nz, 99) * factor) if nz.size else 1.0

def _show(ax, img, channel, label, factor, lim=None):
    # lim=None keeps nb0's behaviour (each panel scaled to itself). Pass an explicit
    # lim to force two panels onto one scale -- required for GT vs PREDICTION, or a
    # prediction that is uniformly 10x too bright looks identical to the truth.
    if lim is None:
        lim = _lim(img, factor)
    if "hmi" not in channel:
        ax.imshow(img, cmap=f"sdo{channel}", vmin=0, vmax=lim)
        fc = "w"
    else:
        fc = "k"
        ax.imshow(img, cmap="coolwarm" if "_v" in channel else "hmimag",
                  vmin=-lim, vmax=lim)
    ax.text(0.01, 0.99, label, transform=ax.transAxes, ha="left", va="top",
            color=fc, fontsize=7)
    ax.set_xticks([]); ax.set_yticks([])

n_out   = 2 * n_lead                                  # GT columns + PRED columns
n_cols  = n_inputs + 1 + n_out                        # +1 = spacer column
w_ratios = [1] * n_inputs + [0.10] + [1] * n_out

for tag, factor in SCALE_VERSIONS:
    n_rows = len(CHANNELS_TO_PLOT) * 2                # image rows + diff rows
    panel = 3.4                                    # inches per (square) panel
    title_in = 0.42                                # inches reserved for suptitle
    fig_w = panel * (n_inputs + n_out) + panel * w_ratios[n_inputs]
    fig_h = panel * n_rows + title_in
    fig = plt.figure(figsize=(fig_w, fig_h), dpi=110)
    gs = gridspec.GridSpec(n_rows, n_cols, figure=fig,
                           width_ratios=w_ratios,
                           wspace=0.02, hspace=0.03,
                           left=0.004, right=0.996,
                           top=1 - title_in / fig_h,   # <- kills the title gap
                           bottom=0.004)
    # ---- image rows: one per channel, columns = time ----
    for r, c in enumerate(CHANNELS_TO_PLOT):
        ins, tgts, prds = frames[c]
        for t in range(n_inputs):
            _show(fig.add_subplot(gs[r, t]), ins[t], c,
                  f"{c}\n{input_times[t]}\nINPUT {t}", factor)
        for l in range(n_lead):
            gt_lim = _lim(tgts[l], factor)
            _show(fig.add_subplot(gs[r, n_inputs + 1 + l]), tgts[l], c,
                  f"{c}\n{target_times[l]}\nGROUND TRUTH", factor)
            _show(fig.add_subplot(gs[r, n_inputs + 1 + n_lead + l]), prds[l], c,
                  f"{c}\n{target_times[l]}\nPREDICTION", factor,
                  lim=gt_lim if SHARE_GT_PRED_SCALE else None)
    # ---- diff rows: centered below, layout of grid above untouched ----
    for r, c in enumerate(CHANNELS_TO_PLOT):
        ins, tgts, prds = frames[c]
        diffs = []
        if n_inputs >= 2:
            diffs.append((ins[-1] - ins[-2],
                          f"{c}  Δ inputs\n{input_times[-1]} − {input_times[-2]}", None))
        # The signal the model has to produce...
        d_signal = tgts[0] - ins[-1]
        lim_sig = np.percentile(np.abs(d_signal), 99) * factor or 1.0
        diffs.append((d_signal,
                      f"{c}  Δ to ground truth\n{target_times[0]} − {input_times[-1]}", lim_sig))
        # ...and how far off it is, on the SAME scale. Same amplitude as the panel to
        # its left => the model captured none of the change.
        diffs.append((prds[0] - tgts[0],
                      f"{c}  Δ prediction error\nPREDICTION − GROUND TRUTH", lim_sig))

        gap_u = w_ratios[n_inputs]
        n_d   = len(diffs)
        side  = (sum(w_ratios) - (n_d + gap_u * (n_d - 1))) / 2
        ratios = [side]
        for k in range(n_d):
            if k:
                ratios.append(gap_u)
            ratios.append(1)
        ratios.append(side)
        inner = gs[len(CHANNELS_TO_PLOT) + r, :].subgridspec(
            1, len(ratios), width_ratios=ratios, wspace=0)
        for k, (d, title, forced) in enumerate(diffs):
            ax = fig.add_subplot(inner[0, 2 * k + 1])
            lim = forced if forced is not None else (np.percentile(np.abs(d), 99) * factor or 1.0)
            im = ax.imshow(d, cmap="coolwarm", vmin=-lim, vmax=lim)
            ax.text(0.01, 0.99, f"{title}\nRMSE {np.sqrt((d**2).mean()):.4g}",
                    transform=ax.transAxes, ha="left", va="top",
                    fontsize=7, color="k",
                    bbox=dict(fc="w", ec="none", alpha=0.65, pad=1.5))
            ax.set_xticks([]); ax.set_yticks([])
            cax = ax.inset_axes([1.015, 0.0, 0.022, 1.0])
            cb = fig.colorbar(im, cax=cax)
            cb.ax.tick_params(labelsize=6)
    fig.suptitle(
        f"[{tag} scaling, 99th pct × {factor}]   {row['GOES_class']}   "
        f"anchor='{anchor_col}' @ {anchor_event}   |   GT frame {gt_time} "
        f"(gap {gap.total_seconds()/60:.0f} min)   |   label={float(item['ground_truth']):.4g}"
        f"   |   MSE {mse_model:.4f} vs persistence {mse_persist:.4f}",
        fontsize=10, y=0.999)
    plt.show()
````
